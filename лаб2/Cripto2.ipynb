```python
import gostcrypto
from sympy import factorint
import time
fio = "Курносов Максим Юрьевич".encode('utf-8')

# Хеширование Стрибог-256
hash_obj = gostcrypto.gosthash.new('streebog256', data=fio)
hash_bytes = hash_obj.digest()

variant_number = hash_bytes[-1]

print(f"Хеш (hex): {hash_bytes.hex()}")
print(f"Номер варианта: {variant_number}")
```

```python



print("Факторизация a с помощью sympy (может занять несколько минут)...")
start = time.time()

try:
    factors = factorint(a, verbose=True)
    elapsed = time.time() - start

    print(f"\nРезультат за {elapsed:.2f} секунд:")
    for prime, exp in factors.items():
        print(f"{prime}^{exp}")

    # Если нашли ровно 2 простых множителя
    if len(factors) == 2:
        p, q = list(factors.keys())
        print(f"\np = {p}")
        print(f"q = {q}")
        print(f"p * q = {p * q}")

except Exception as e:
    print(f"Ошибка: {e}")
```
